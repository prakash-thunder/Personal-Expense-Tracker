# 🚀 Project 1: Personal Expense Tracker

## 📌 Project Overview

A beginner-friendly Python project to build a **Personal Expense Tracker**.

The program will allow users to record their daily expenses and analyze their spending.

---

## 🎯 Project Goal

Build a Python program that can:

- Add a new expense
- Store the date of the expense
- Store the expense category
- Store the amount
- View all expenses
- Calculate total spending
- Find the highest expense
- Find spending by category
- Save expenses to a CSV file

````python
from datetime import date,datetime
expenses={}
def expense_on_item():
  item=input("Enter item name like grocery,books,other: ").lower().strip()
  amount=float(input('enter amount: '))
  a=input("Enter 'today' or enter date in format dd/mm/yyyy: ").lower()
  if a=='today':
    date_=date.today()
  else:
    date_ = datetime.strptime(a, "%d/%m/%Y").date()
  if not expenses.get(date_):
    expenses[date_]={}
    data=expenses[date_]
    data[str(item)]=data.get(str(item),0)+amount
    return
  data=expenses[date_]
  data[str(item)]=data.get(str(item),0)+amount
````

````python
def total_expenses_a_day():
  a=input("Enter 'today' or enter date in format dd/mm/yyyy: ").lower()
  if a=='today':
    date_=date.today()
  else:
    date_ = datetime.strptime(a, "%d/%m/%Y").date()
  if expenses.get(date_) :
    data=expenses[date_]
    total=0
    for _,i in data.items():
      total+=i
    return total
  return 0
````

````python
def total_expenses():
  total=0
  for i in expenses:
    data=expenses[i]
    for _,i in data.items():
      total+=i
  return total
````

````python
def spending_by_item():
  item=input("Enter item name like grocery,books,other")
  total=0
  for i in expenses:
    data=expenses[i]
    total+=data.get(item,0)
  print("total amount spend on this item :")
  return total
````

````python
def find_highest_expenses():
  highest_amount=0
  highest_day_total=0
  for i in expenses:
    total=0
    data=expenses[i]
    for it,j in data.items():
      total+=j
      if highest_amount<j:
        highest_amount=max(highest_amount,j)
        item=it
        date_=i
    if highest_day_total<total:
      s_date=i
      highest_day_total=max(highest_day_total,total)
  return f'Highest individual expense: {item} - ₹{highest_amount} on {date_.strftime("%d/%m/%Y")} \n Highest spending day: {s_date.strftime("%d/%m/%Y")} - ₹{highest_day_total}'
````

````python
find_highest_expenses()
````
